```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

def audit_website(url):
    try:
        start_time = time.time()
        # Website data fetch karne ke liye (Google bot jaisa behavior simulate karne ke liye User-Agent lagaya hai)
        response = requests.get(url, headers={'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'}, timeout=10)
        latency = time.time() - start_time

        soup = BeautifulSoup(response.text, 'html.parser')

        # SEO Elements Extract karna
        title = soup.find('title').text.strip() if soup.find('title') else "Missing Title"

        meta_desc = "Missing Meta Description"
        meta_tag = soup.find('meta', attrs={'name': 'description'})
        if meta_tag:
            meta_desc = meta_tag.get('content', '').strip()

        h1_tags = [h1.text.strip() for h1 in soup.find_all('h1')]
        h1_count = len(h1_tags)

        return {
            "URL": url,
            "Status Code": response.status_code,
            "Title": title,
            "Title Length": len(title),
            "Meta Description": meta_desc,
            "Meta Length": len(meta_desc),
            "H1 Count": h1_count,
            "H1 Content": ", ".join(h1_tags) if h1_tags else "None",
            "Page Speed Latency (secs)": round(latency, 2)
        }
    except Exception as e:
        return {
            "URL": url, "Status Code": "Error", "Title": "N/A", "Title Length": 0,
            "Meta Description": "N/A", "Meta Length": 0, "H1 Count": 0, "H1 Content": "N/A",
            "Page Speed Latency (secs)": 0.0
        }

# YAHAN APNE TARGET URLS DAALEIN
urls_to_test = [
    "https://example.com",
    "https://httpbin.org",
    "https://wikipedia.org"
]

print("SEO Audit shuru ho raha hai...")
results = [audit_website(u) for u in urls_to_test]

# Data ko Excel/CSV mein convert karna
df = pd.DataFrame(results)
df.to_csv("seo_audit_report.csv", index=False)
print("Audit Complete! 'seo_audit_report.csv' file ban gayi hai.")
```
